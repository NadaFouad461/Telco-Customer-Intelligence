```python
import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_excel("E:\MachineLearning\Data\Telco_customer_churn.xlsx")
```

```python
df.info()
```

```python
df["Total Charges"] = (
    df["Total Charges"]
    .astype(str)
    .str.strip()
    .replace("", np.nan)
)

df["Total Charges"] = pd.to_numeric(
    df["Total Charges"],
    errors="coerce"
)
```

```python
df["Total Charges"].isna().sum()
```

```python
df.loc[
    df["Total Charges"].isna(),
    ["CustomerID", "Tenure Months", "Monthly Charges", "Churn Label"]
]
```

```python
df.loc[df["Total Charges"].isna(), "Total Charges"] = 0
```

```python
df["Total Charges"].isna().sum()
```

```python
df["Total Charges"].dtype
```

```python
df = df.drop(columns=["Count", "Country", "State"])
```
