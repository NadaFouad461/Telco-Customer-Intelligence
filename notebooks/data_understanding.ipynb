# Data Understanding Summary

## Dataset Overview

- Number of rows: 7043
- Number of columns: 33
- Duplicate rows: 0
- Unique customers: 7043

## Missing Values

- `Churn Reason` contains 5174 missing values.
- Missingness is associated with customers who did not churn.
- Other columns do not contain missing values at this stage.

## Data Quality Observations

- `Count` is constant with value 1 for all customers.
- `Country` is constant (`United States`).
- `State` is constant (`California`).
- `Total Charges` is stored as an object and contains blank string values.
- `CustomerID` is unique for every customer.

## Target Variables

### Classification Target
`Churn Label`

### Regression Target
`CLTV`

## Potential Leakage / Excluded Features

- `Churn Value`: duplicate representation of `Churn Label`.
- `Churn Score`: existing churn-related score; should not be used as an input feature for the baseline churn prediction model.
- `Churn Reason`: information available after/associated with churn and therefore excluded from churn prediction features.
- `CustomerID`: identifier, not a predictive feature.
